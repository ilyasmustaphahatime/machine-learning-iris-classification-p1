```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42,
    stratify=Y
)

training_data = X_train.copy()
training_data["class"] = Y_train

print(training_data.head())
```

```python
print("Original flower:")
print(X.loc[31])

print("\nFlower after splitting:")
if 31 in X_train.index:
    print(X_train.loc[31])
else:
    print(X_test.loc[31])
```

```python
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, Y_train)
print("Model trained successfully!")
```

```python
Y_pred = model.predict(X_test)
print(Y_pred[:10])

```

```python
print("Predictions:")
print(Y_pred[:10])

print("\nActual classes:")
print(Y_test.iloc[:10].values)
```

```python
# calcualting of the accuracy and comparin between testing target anad the predict one 
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(Y_test, Y_pred)

print("Accuracy:", accuracy)
print(f"Accuracy: {accuracy:.2%}")
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    Y_test,
    Y_pred,
    labels=["Iris-setosa", "Iris-versicolor", "Iris-virginica"]
)

print(cm)
```

```python
import pandas as pd

labels = [
    "Iris-setosa",
    "Iris-versicolor",
    "Iris-virginica"
]

cm_table = pd.DataFrame(
    cm,
    index=labels,
    columns=labels
)

print(cm_table)
```

```python
import pandas as pd

results = pd.DataFrame({
    "Actual": Y_test.values,
    "Predicted": Y_pred
})

results["Correct"] = results["Actual"] == results["Predicted"]

print(results.head(10))
```

```python
#creating a table for checking the accuracy and the recall and f1
import pandas as pd

labels = [
    "Iris-setosa",
    "Iris-versicolor",
    "Iris-virginica"
]

cm_table = pd.DataFrame(
    cm,
    index=labels,
    columns=labels
)

print(cm_table)
```

```python
for i, flower in enumerate(labels):
    TP = cm[i, i]
    FP = cm[:, i].sum() - TP
    FN = cm[i, :].sum() - TP
    TN = cm.sum() - TP - FP - FN

    print(f"\n{flower}")
    print("TP:", TP)
    print("FP:", FP)
    print("FN:", FN)
    print("TN:", TN)
```
